# BFR Analysis Data Exploration

```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
```

## Subject 2 no BFR

### Find SA MVC

```python
# SA muscle
df_mvc = pd.read_csv("output/PB TT Tennis no BFR ALL TRIALs/Subject 2 no BFR/Subject_2_KB__SA_MVC_(EMG_RMS)_(2)_Rep_1.1.hpf-ch_data.csv")

print(df_mvc.columns)

df_mvc.head()
```

```python
sa_emg_raw = df_mvc['Serratus Anterior 2: EMG 2 [V]']
print("Max: ", max(sa_emg_raw), "| Min: ", min(sa_emg_raw))
sa_emg_raw_fig = go.Figure(
    data=go.Scatter(
        x=df_mvc["X [s]"], 
        y=df_mvc["Serratus Anterior 2: EMG 2 [V]"], 
        name='Raw data', 
        line=dict(color='blue')
    )
)
sa_emg_raw_fig.update_layout(
    title=dict(
        text='Subject 2 - SA - EMG - Raw data'
    ),
    xaxis=dict(
        title=dict(
            text='Time [S]'
        )
    ),
    yaxis=dict(
        title=dict(
            text='Serratus Anterior 2: EMG 2 [V]'
        )
    ),
)
sa_emg_raw_fig.show()
```

#### RMS Smoothing
> Reference: https://delsys.com/amplitude-analysis-root-mean-square-emg-envelope/

```python
# Smoothing function
def rms_smoothing(df: pd.DataFrame, raw_column: str, smoothed_column: str, window_size: int):
    df[smoothed_column] = np.sqrt(
        df[raw_column].pow(2).rolling(window=window_size).mean()
    )

window_size = 10
rms_smoothing(df_mvc, "Serratus Anterior 2: EMG 2 [V]", "rms_smoothed", window_size)

sa_emg_rms = df_mvc['rms_smoothed']
print("Max: ", np.max(sa_emg_rms), "| Min: ", np.min(sa_emg_rms))
sa_emg_rms_fig = go.Figure(
    data=go.Scatter(
        x=df_mvc["X [s]"], 
        y=df_mvc["rms_smoothed"], 
        name='RMS Smoothed', 
        line=dict(color='red')
    )
)
sa_emg_rms_fig.update_layout(
    title=dict(
        text='Subject 2 - SA - EMG - RMS Smoothed'
    ),
    xaxis=dict(
        title=dict(
            text='Time [S]'
        )
    ),
    yaxis=dict(
        title=dict(
            text='Serratus Anterior 2: EMG 2 RMS Smoothed [V]'
        )
    ),
)
sa_emg_rms_fig.show()
```

```python
sa_emg_fig = go.Figure()

sa_emg_fig.add_trace(
  go.Scatter(
        x=df_mvc["X [s]"], 
        y=df_mvc["Serratus Anterior 2: EMG 2 [V]"], 
        name='Raw data', 
        line=dict(color='blue')
  )
)
sa_emg_fig.add_trace(
    go.Scatter(
        x=df_mvc["X [s]"], 
        y=df_mvc["rms_smoothed"], 
        name='RMS Smoothed', 
        line=dict(color='red')
    )
)
sa_emg_fig.update_layout(
    title=dict(
        text='Subject 2 - SA - EMG - Combined'
    ),
    xaxis=dict(
        title=dict(
            text='Time [S]'
        )
    ),
    yaxis=dict(
        title=dict(
            text='Serratus Anterior 2: EMG 2 Combined [V]'
        )
    ),
)
sa_emg_fig.show()
```

### Normalization - Pickleball BH - Rep 3 - SA

```python
df_pkbh = pd.read_csv("./output/PB TT Tennis no BFR ALL TRIALs/Subject 2 no BFR/Subject_2_KB_Pickleball_BH_Rep_3_Rep_1.23.hpf-ch_data.csv")
print("SA - Max: ", max(df_pkbh["Serratus Anterior 2: EMG 2 [V]"]), "| Min: ", min(df_pkbh["Serratus Anterior 2: EMG 2 [V]"]))
df_pkbh.head()
```

#### Apply the Same RMS Smoothing

```python
# Using the previously defined smoothing function and window size
rms_smoothing(df_pkbh, "Serratus Anterior 2: EMG 2 [V]", "smoothed_sa", window_size)

pkbh_sa_emg_fig = go.Figure()

pkbh_sa_emg_fig.add_trace(
  go.Scatter(
        x=df_pkbh["X [s]"], 
        y=df_pkbh["Serratus Anterior 2: EMG 2 [V]"], 
        name='Raw data', 
        line=dict(color='blue')
  )
)
pkbh_sa_emg_fig.add_trace(
    go.Scatter(
        x=df_pkbh["X [s]"], 
        y=df_pkbh["smoothed_sa"], 
        name='RMS Smoothed', 
        line=dict(color='red')
    )
)
pkbh_sa_emg_fig.update_layout(
    title=dict(
        text='Subject 2 - Pickleball BH - Rep3 - SA - EMG - Combined'
    ),
    xaxis=dict(
        title=dict(
            text='Time [S]'
        )
    ),
    yaxis=dict(
        title=dict(
            text='Serratus Anterior 2: EMG 2 Combined [V]'
        )
    ),
)
pkbh_sa_emg_fig.show()
```

```python
# previously got subject 2 SA MVC RMS Smoothed Max
mvc_sa_max = np.max(df_mvc["rms_smoothed"])
print("Smoothed MVC Max: ", mvc_sa_max)

# normalized = SA RMS Smoothed / SA MVC Smoothed Max * 100%
df_pkbh["sa_norm_percent"] = df_pkbh["smoothed_sa"].apply(lambda x: np.divide(x, mvc_sa_max)*100)
np.max(df_pkbh["sa_norm_percent"])


pkbh_sa_emg_norm_fig = go.Figure()

pkbh_sa_emg_norm_fig.add_trace(
  go.Scatter(
        x=df_pkbh["X [s]"], 
        y=df_pkbh["sa_norm_percent"], 
        name='Smoothed then Normalized SA EMG', 
        line=dict(color='green')
  )
)
pkbh_sa_emg_norm_fig.update_layout(
    title=dict(
        text='Subject 2 - Pickleball BH - Rep3 - SA - EMG - Smoothed then Normalized'
    ),
    xaxis=dict(
        title=dict(
            text='Time [S]'
        )
    ),
    yaxis=dict(
        title=dict(
            text='Percentage of Smoothed MVC Max'
        )
    ),
)
pkbh_sa_emg_norm_fig.show()
```
